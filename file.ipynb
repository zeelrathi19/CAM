```python
import cv2
import json
import numpy as np
from ultralytics import YOLO  # pip install ultralytics

# === Load L-shape points from JSON ===
with open("lshape_cam.json", "r") as f:
    l_points = json.load(f)

A, B, C = [np.array(p) for p in l_points]
line1 = (A, B)
line2 = (B, C)

# === Load YOLOv8 model ===
model_path = "yolo11n.pt"  # your custom model
model = YOLO(model_path)
print(f"✅ YOLO model loaded from: {model_path}")

# === Open RTSP Camera ===
CAMERA_INDEX = "rtsp://admin:team%401703@103.142.162.188:554/cam/realmonitor?channel=1&subtype=0"
cap = cv2.VideoCapture(CAMERA_INDEX, cv2.CAP_FFMPEG)

if not cap.isOpened():
    print("❌ Failed to open camera stream.")
    exit()

# === Function to determine side of a point relative to a line ===
def point_side(pt, line_start, line_end):
    # Convert to 3D by adding a zero z-coordinate
    v1 = np.append(line_end - line_start, 0)
    v2 = np.append(pt - line_start, 0)
    cross_prod = np.cross(v1, v2)
    return cross_prod[2]  # Return z component (scalar in 2D)


while True:
    ret, frame = cap.read()
    if not ret:
        print("❌ Failed to grab frame.")
        break

    # === Run YOLOv8 Inference ===
    results = model.predict(frame, verbose=False)[0]

    for person in results.boxes:
        cls_id = int(person.cls[0])
        label = model.names[cls_id]

        if "person" in label.lower():  
            x1, y1, x2, y2 = map(int, person.xyxy[0])
            bottom_left = np.array([x1, y2])
            bottom_right = np.array([x2, y2])

            # Check if either bottom corner is outside the L-shape
            out_L = (
                point_side(bottom_left, *line1) > 0 or point_side(bottom_left, *line2) < 0 or
                point_side(bottom_right, *line1) > 0 or point_side(bottom_right, *line2) < 0
            )

            color = (0, 0, 255) if out_L else (0, 255, 0)
            status = "⚠️ Outside L" if out_L else "✅ Inside L"

            # Draw box and label
            cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
            cv2.putText(frame, status, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)

    # === Draw L-shape ===
    cv2.line(frame, tuple(A), tuple(B), (0, 255, 255), 2)
    cv2.line(frame, tuple(B), tuple(C), (0, 255, 255), 2)

    # Mark points A, B, C
    cv2.circle(frame, tuple(A), 5, (0, 0, 255), -1)
    cv2.circle(frame, tuple(B), 5, (0, 255, 0), -1)
    cv2.circle(frame, tuple(C), 5, (255, 0, 0), -1)
    cv2.putText(frame, "A", tuple(A), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 1)
    cv2.putText(frame, "B", tuple(B), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)
    cv2.putText(frame, "C", tuple(C), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 0, 0), 1)

    # === Show Frame ===
    cv2.imshow("person Detection with L-Shape", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
